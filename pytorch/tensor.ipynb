# 3. 인덱싱과 슬라이싱

```python
t = torch.arange(24).reshape(2, 3, 4) # 0~23까지의 숫자로 이루어진 텐서를 reshape를 통해서 3D-텐서로 바꿈
print(t.shape)
print(t)
```

```python
t[0].shape # (3, 4) 첫 번째 axis 에서 하나를 선택 -> 해당 축이 사라진다고 생각하면 됨
```

```python
t[:, 0].shape # (2, 4) 두 번째 축에서 0번째 애만 선택 -> 해당 축이 사라지고 : 은 해당 축의 전체를 선택하는 것을 의미함. 첫 번째 축을 명시적으로 표현하는 것도 두 번째 축을 명시하기 위해서임
```

```python
t[..., -1].shape # (2, 3) ..., 은 앞에 있는 축 전부를 뜻함
```

```python
t[:, 1:3].shape # (2, 2, 4)
```

```python
b = torch.arange(6)
b[::2] # 자리 0부터 두 칸씩 건너뛰면서 선택
```

```python
b[1::2] # 자리 1부터 두칸씩
```

# 4. 텐서의 모양 바꾸기
- 앞에서 다룬 reshape

```python
x = torch.arange(12) # 12차원의 1D 텐서
x.reshape(3, 4).shape # 3행 4열의 2D 텐서로 모양이 바뀐다
```

```python
x = torch.arange(12)
x.reshape(3, -1).shape # 첫 번째 축의 크기를 정해주면 -1로 나머지는 알아서 계산된다는 의미이다.
```

```python
img = torch.rand(8, 3, 32, 32) # 보통 이미지 데이터는 다음과 같이 4D 텐서이다 (B, C, H, W) 배치, 채널, 높이, 너비
```

```python
img.flatten(1).shape # (8, 3072) 1번축부터 끝까지 하나로 펼치는 작업
```

```python
img.permute(0, 2, 3, 1).shape # 축의 순서를 바꾸는 작업. 1번 축을 제일 뒤로 보낸다.
```

```python
v = torch.rand(5) # (5,)
```

```python
v.unsqueeze(0).shape # (1, 5) 지정한 위치에 크기가 1인 축을 추가한다.
```

```python
v.unsqueeze(1).shape # (5, 1)
```

```python
v[None, :].shape # unsqueeze(0)과 같은 동작을 한다.
```

```python
torch.rand(1, 5, 1).squeeze().shape # (5, ) 크기가 1인 축을 전부 제거한다
```

reshape vs view
- 둘 다 tensor의 모양을 바꾼다
- 그러나 view는 메모리가 연속으로 놓여있어야만 동작한다
    - 전치(transpose; .T)나 permute를 하고 나면 메모리 배치가 바뀌기 때문에 view가 실패할 수 있음

```python
t = torch.rand(3, 4).T
t.is_contiguous() # False
# t.view(12) # 에러
# RuntimeError: view size is not compatible with input tensor's size and stride (at least one dimension spans across two contiguous subspaces). Use .reshape(...) instead.
```

```python
t.reshape(12) # 동작, 필요시 알아서 복사해서 진행됨
```

# 5. 합치기: cat vs stack

```python
p = torch.zeros(2, 3)
q = torch.ones(2, 3)
```

```python
a = torch.cat([p, q], dim=0) # (4, 3) 기존 0번 축을 따라서 이어 붙인다. 차원 수가 유지됨
a.shape
```

```python
print(a)
```

```python
b = torch.cat([p, q], dim=1)
b.shape # (2, 6)
```

```python
print(b)
```

```python
c = torch.stack([p, q], dim=0)
c.shape # (2, 2, 3) 새로운 축을 하나 만들어서 쌓는다
```

```python
print(c)
```

# 7. 축 줄이기

```python
s
```

```python
s.sum(dim=0).shape # (3,) 0번축이 사라진다
```

```python
s.sum(dim=0)
```

```python
s.sum(dim=1).shape # (2,)
```

```python
s.sum(dim=1)
```

```python
s.sum(dim=1, keepdim=True).shape #(2, 1) 축을 지우지 않고 크기 1로 남긴다
```

```python
s.sum(dim=1, keepdim=True)
```

`dim=k`는 k번 축을 따라 연산해서 그 축을 없앤다로 이해하자
- `mean`, `max`, `argmin` 도 위와 같은 규칙을 따름

# 8. 브로드캐스팅(broadcasting)
- 오른쪽 끝 차원부터 짝지어 비교한다
- 짝지은 두 차원의 크기가 같거나 둘 중 하나가 1이면 가능하다. 1인 쪽이 상대 크기만큼 늘어난다.(브로드캐스팅)
- 차원 수 자체가 모자란 경우 앞쪽에 1이 붙은 것으로 취급한다.

```python
(torch.ones(2, 3) + torch.ones(3)).shape # (2, 3) : (3,) -> (1, 3) -> (2, 3)
```

```python
(torch.ones(2, 1) + torch.ones(1, 3)).shape # (2, 3)
# (2, 1) -> (2, 3)
# (1, 3) -> (2, 3)
```

```python
(torch.ones(4, 1) + torch.ones(5)).shape # (4, 5)
# (4, 1) -> (4, 5)
# (5,) -> (1, 5) -> (4, 5)
```
